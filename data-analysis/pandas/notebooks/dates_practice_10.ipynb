# Topic 10 practice: dates

The tables are built in code, so no CSV is needed. Re-run the setup cell to reset them. Start each question from `trips`, `deliveries` or `bills`, work on a copy, and give each result a new name.

Tip: a comment line must start with `#`.

```python
import pandas as pd

trips = pd.DataFrame({"trip":   [1, 2, 3, 4, 5, 6],
                      "start":  ["2026-10-01 09:30", "2026-10-05 18:45", "2026-09-28 07:10",
                                 "2026-10-12 14:00", "2026-09-13 22:15", "2026-11-02 11:20"],
                      "amount": [500, 300, 450, 800, 250, 600]})

deliveries = pd.DataFrame({"order_id":  [1, 2, 3, 4, 5],
                           "ordered":   ["2026-10-01", "2026-10-03", "2026-09-25", "2026-10-10", "2026-09-30"],
                           "delivered": ["2026-10-04", "2026-10-10", "2026-09-27", None, "2026-10-02"],
                           "amount":    [500, 300, 450, 800, 250]})

bills = pd.DataFrame({"bill_id": [1, 2, 3, 4, 5, 6],
                      "date":    ["15/10/2026", "03/10/2026", "28/09/2026", "12/11/2026", "20/09/2026", "05/10/2026"],
                      "amount":  [400, 250, 700, 300, 150, 500]})
print(trips)
print(deliveries)
print(bills)
```

### Q1 (easy). Text to dates and parts

In `trips`, `start` is text. Make a copy `t1` and turn `start` into real dates. Add three columns: `month` (month number), `weekday` (day name) and `is_weekend` (True for Saturday or Sunday). Print `t1`. Then print the list of trip numbers that happened in October.

```python
# your code here (add a comment saying why for each line)
t1 = trips.copy()
t1["start"] = pd.to_datetime(t1["start"])
t1["months"] = t1["start"].dt.month
t1["weekday"] = t1["start"].dt.day_name()
t1["is_weekend"] = t1["start"].dt.dayofweek >= 5
print(t1)
print(t1[t1["months"] == 10]["trip"].tolist())  
```

### Q2 (medium). Days between dates

In `deliveries`, make a copy `d2` and turn `ordered` and `delivered` into real dates. Add `days_taken` (days from ordered to delivered) and `due` (the order date plus 7 days). Print `d2`. Then print the average of `days_taken`, the `order_id` list that took more than 3 days, and the `order_id` list not delivered yet.

```python
# your code here (add a comment saying why for each line)
dr = deliveries.copy()
dr["delivered"] = pd.to_datetime(dr["delivered"])
dr["ordered"] = pd.to_datetime(dr["ordered"])
dr["days_taken"] =  dr["delivered"] - dr["ordered"] 
dr["due"] = dr["ordered"] + pd.Timedelta(days=7)
print(dr)
print(dr["days_taken"].mean())

```

### Q3 (medium to hard). Day-first dates and months

In `bills`, the date is written day first, like 15/10/2026. Make a copy `b3` and turn `date` into real dates (day first!). Print the total `amount` for each month (the month as 2026-09, 2026-10 ...). Then print the month with the biggest total and the `bill_id` of the newest bill.

```python
# your code here (add a comment saying why for each line)
b3 = bills.copy()
b3["date"] = pd.to_datetime(b3["date"], format="%d/%m/%Y")
monthly = b3.groupby(b3["date"].dt.to_period("M"))["amount"].sum()
print(monthly)
print(monthly.idxmax())
print(b3.sort_values("date", ascending=False).iloc[0]["bill_id"])
```
