# Topic 7 practice: sorting

The tables are built in code, so no CSV is needed. Re-run the setup cell to reset them. Start each question from `students`, `exam` or `shop`, and give each result a new name.

Tip: a comment line must start with `#`.

```python
import pandas as pd

students = pd.DataFrame({"name": ["Asha", "Vivek", "Ravi", "Meena"],
                         "city": ["Indore", "Pune", "Delhi", "Indore"],
                         "marks": [82, 91, 67, 75]})

exam = pd.DataFrame({"name": ["Asha", "Vivek", "Ravi", "Meena", "Kiran", "Dev"],
                     "city": ["Indore", "Pune", "Delhi", "Indore", "Pune", "Delhi"],
                     "marks": [82, 91, 67, 82, 91, 67]})

shop = pd.DataFrame({"item": ["pen", "doll", "novel", "rice", "ball"],
                     "price": [10, 250, 400, 60, 120],
                     "rating": [4.5, None, 3.8, None, 4.2]})
print(students)
print(exam)
print(shop)
```

### Q1 (easy). Sort high to low

Sort `students` by `marks`, highest first. Save in `s1`. Print `s1` and then the list of names in that order.

```python
# your code here (add a comment saying why for each line)
s = students.copy()
s1 = s.sort_values("marks", ascending=False)
list(s1["name"])
```

### Q2 (easy). Sort text and renumber

Sort `students` by `name`, A to Z, with fresh labels 0 to 3. Save in `s2` and print it. Then print `students.index.tolist()` to show the original did not change.

```python
# your code here (add a comment saying why for each line)
s = students.copy()
s1 = s.sort_values("name", ignore_index=True)
print(s1)
print(students.index.tolist())
```

### Q3 (easy to medium). Top two

Get the names of the top 2 students by marks in two ways: sort then `head`, and `nlargest`. Print both name lists and whether they are equal.

```python
# your code here (add a comment saying why for each line)
s = students.copy()
a = s.sort_values("marks",ascending=False).head(2)
b = s.nlargest(2,"marks")
print(a["name"].tolist())                                             # ['Vivek', 'Asha']
print(b["name"].tolist())  

print(a.equals(b))
```

### Q4 (medium). Two columns

Sort `exam` by `marks` (highest first) and, when marks are the same, by `name` A to Z. Save in `s4` and print the list of names in that order.

```python
# your code here (add a comment saying why for each line)
e = exam.copy()
a = e.sort_values(["marks","name"] , ascending=[False, True])
print(list(a["name"]))
```

### Q5 (medium). Missing values

Sort `shop` by `rating`, highest first. Print the items in that order and say in a comment where the missing ratings land. Then print the items again with the missing ratings placed first. Finally keep only the rows that have a rating and sort them highest first.

```python
# your code here (add a comment saying why for each line)
s = shop.copy()
s1 = s.sort_values("rating", ascending=False)
print(s1)
print(list(s1["item"]))

s2 = s.sort_values("rating", ascending=False, na_position="first" ) 
clean = s.dropna(subset=["rating"]).sort_values("rating", ascending=False)       # the option that puts NaN on top
```

### Q6 (medium to hard). Labels after sorting

Sort `students` by `marks` (low to high) into `r`. Print `r.index.tolist()`. Then restore the original order using the labels, and check with `equals` that it is the same as `students`. Last, set `name` as the index and print the table sorted by name Z to A.

```python
# your code here (add a comment saying why for each line)
s = students.copy()
r = s.sort_values("marks")
print(r.index.tolist())
r1 = r.sort_index()
print(r1)
print(s.equals(r1))
```

### Q7 (hard). Rank

1) Add a column `rank` to a copy of `students` where the highest marks get rank 1 (whole numbers). Print it, still in the original row order. 2) Sort that table by `rank` and print the names in rank order. 3) In `shop`, drop the rows with no rating, rank the rest by `price` (most expensive = 1, whole numbers), sort by that rank, and print `item` and `rank`.

```python
st = students.copy()                                                  # the students table
st["rank"] = st["marks"].rank(ascending=False).astype(int)            # rank of the marks column, highest = 1
print(st)
print(st.sort_values("rank")["name"].tolist())                        # sort the same table st, not r

sh = shop.dropna(subset=["rating"]).copy()                            # the shop table, only rows that have a rating
sh["rank"] = sh["price"].rank(ascending=False).astype(int)            # most expensive = 1
print(sh.sort_values("rank")[["item", "rank"]])
```
