### Block 1 — Building a DataFrame from a dict
- A list holds values in order inside square brackets: [a, b, c].
- A dict holds key → value pairs inside curly brackets: {"key": value}.
- Look up a dict value by its key: d["key"].
- `pd.DataFrame(dict_of_lists)` builds a table: each key becomes a column name, each list fills that column.
- Item positions line up into rows: the first item of every list forms row 0.
- All lists must have the same length, otherwise pandas raises an error.
- DataFrame is written with capital D and F.
- Building tables from dicts is the quickest way to make small test data without a file.

### Block 2 — What a Series is
- A Series is a single column: values + row labels (index) + an optional name + one dtype.
- A DataFrame is several Series side by side that share the same row labels.
- `pd.Series(list, name="...")` builds a Series from a list.
- A Series is displayed without table borders, with a "Name: ..., dtype: ..." line at the bottom.
- `type(x)` shows whether x is a Series, a DataFrame, or something else.
- Series is written with a capital S.
- In ML, the target column you want to predict is usually a Series.

### Block 3 — Selecting one column
- `df["col"]` selects one column and returns a Series; it always works.
- `df.col` is a shortcut that returns the same Series.
- The dot shortcut fails when the name has a space or matches a pandas method (e.g. count); use brackets then.
- Column names must match exactly, including capital letters and spaces.
- KeyError means no column has that exact name; check spelling, capitals and hidden spaces.
- Prefer brackets: they work for every column name.
