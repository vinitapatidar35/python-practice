### Block 1 — Building a DataFrame from a dict
- A list holds values in order inside square brackets: [a, b, c].
- A dict holds key → value pairs inside curly brackets: {"key": value}.
- Look up a dict value by its key: d["key"].
- `pd.DataFrame(dict_of_lists)` builds a table: each key becomes a column name, each list fills that column.
- Item positions line up into rows: the first item of every list forms row 0.
- All lists must have the same length, otherwise pandas raises an error.
- DataFrame is written with capital D and F.
- Building tables from dicts is the quickest way to make small test data without a file.

### Block 2 — What a Series is
- A Series is a single column: values + row labels (index) + an optional name + one dtype.
- A DataFrame is several Series side by side that share the same row labels.
- `pd.Series(list, name="...")` builds a Series from a list.
- A Series is displayed without table borders, with a "Name: ..., dtype: ..." line at the bottom.
- `type(x)` shows whether x is a Series, a DataFrame, or something else.
- Series is written with a capital S.
- In ML, the target column you want to predict is usually a Series.

### Block 4 — Selecting many columns
- `df[["a", "b"]]` selects several columns: outer brackets = select, inner brackets = list of names.
- The result is a DataFrame, with columns in the order listed.
- The list can be stored in a variable first: `cols = ["a", "b"]`, then `df[cols]`.
- `df["a"]` gives a Series; `df[["a"]]` gives a one-column DataFrame.
- Forgetting the inner brackets (`df["a", "b"]`) raises a KeyError.
- In ML, the chosen input columns (features) are usually selected this way.

### Block 5 — The columns attribute

- Check `df.columns` first whenever a KeyError appears.

### Block 6 — iloc (selecting by position)
- `df.iloc[...]` selects by position number, counting from 0; it uses square brackets.
- `iloc[n]` returns one row as a Series (column names become its labels).
- `iloc[-1]` is the last row; negative positions count from the end.
- `iloc[[a, b]]` returns the listed rows as a DataFrame.
- `iloc[start:stop]` returns a range of rows; the stop position is not included.
- `iloc[row, col]` returns a single value; the row comes first, then the column.
- A position beyond the last row raises IndexError (out-of-bounds).

### Block 7 — loc (selecting by label)
- `df.loc[...]` selects by label: the row labels on the left and the column names.
- `loc[label]` returns one row; `loc[label, "col"]` returns one value.
- `loc[[a, b], ["col1", "col2"]]` returns chosen rows and chosen columns as a DataFrame.
- `loc[start:stop]` includes the stop label (unlike iloc, which excludes it).
- With the default index, labels equal positions; after filtering, sorting or tail(), they differ.
- A label that does not exist raises KeyError.
- iloc = positions and column numbers; loc = labels and column names.

### value_counts()

- Counts how many times each value appears in ONE column (a tally chart)
- Call it on a Series: `df["col"].value_counts()`
- Result is a Series, sorted from biggest count to smallest
- `normalize=True` gives shares (0 to 1); multiply by 100 for percentages
- `dropna=False` also counts missing values (NaN); by default they are hidden
- Argument names a column, not a value inside the column
- Can be chained: `.head(n)` for top n rows, `.index[0]` for the top value's name

Write code to count how many students live in each city.


```python
student = pd.read_csv("students.csv")
print(student["city"].value_counts())
print(student["city"].value_counts(normalize=True))
print((student["city"].value_counts(normalize=True))*100)

print(student.value_counts(dropna=False))   # dropna=False keeps NaN in the count
```

### dtypes

- dtype = the kind of data a column holds; one column has one dtype
- `df.dtypes` shows every column's type (attribute, no brackets); `df["col"].dtype` shows one
- `int64` whole numbers, `float64` decimals, `object` text, `bool` True/False
- What you can do with a column depends on its dtype (add numbers, not text)
- One text cell in a numeric column turns the whole column into `object`
- A column with NaN is stored as `float64`
- `info()` also lists each column's dtype

Q1. Why is qty object and not int64?
One cell in the column is the text "five" (the Bottle row). A column holds only one dtype, and numbers and text cannot share int64. So pandas stores the whole column as object. The other numbers (50, 20, 12, 3, 1) are no longer treated as numbers either, so you cannot add or average them until the bad cell is fixed.

Q2. Why is rating float64?
Two cells in the column are empty (Notebook and Lamp), and pandas stores an empty cell as NaN. NaN is a decimal (float) value, and a whole column cannot be int64 while holding it. The other values are converted to decimals too. So the column is float64 because of the NaN, not because of the 4.5.
