```python
import pandas as pd                                    # open the pandas toolbox
pd.set_option("display.max_columns", None)             # show every column in every table

students = pd.read_csv("students.csv")                 # name, city, marks (4 rows)
orders = pd.read_csv("orders.csv")                     # order_id, customer, city, amount, items (8 rows)
shop = pd.read_csv("shop.csv")                         # item, category, price, qty, rating (6 rows)
```

### Q1 (easy). On `students`: build a mask `passed` that is True where marks are 75 or more. Print the mask, then print the filtered table, then print how many rows passed using `len()`.

```python
mask = students["marks"] >= 75
print(mask)
print(students[mask])
print(len(students[mask]))
```

### Q2 (easy). On `orders`: print the rows from Pune or Delhi using `isin`, then the rows that are NOT from Pune or Delhi. Say in a comment how many rows each gives.

```python
order = pd.read_csv("orders.csv")
print(order["city"].isin(["Pune","Delhi"]))
print(len(order["city"].isin(["Pune","Delhi"])))
print(~(order["city"].isin(["Pune","Delhi"])))
print(len(~(order["city"].isin(["Pune","Delhi"]))))
```

```python
mask = orders["city"].isin(["Pune", "Delhi"])     # True/False per order
print(orders[mask])                               # the rows from Pune or Delhi
print(len(orders[mask]))                          # 4
print(orders[~mask])                              # the rest
print(len(orders[~mask]))                         # 4
print(mask.sum())                                 # 4: another way, True counts as 1 (same trick as isna().sum())
```

### Q3 (medium). On `shop`: print the item names whose price is from 100 to 900, both ends included, using `between`. Then print the same names with `loc` (rows from the mask, one column).

```python
name = shop["price"].between(100,900)
print(name)
print(shop[name])
print(len(shop[name]))
print(shop.loc[name,"item"])
```

```python
mid = shop["price"].between(100, 900)         # a clearer name than "name": it is a mask, not a name
print(shop[mid]["item"].tolist())             # ['Bottle', 'Mug', 'Lamp']
print(shop.loc[mid, "item"])                  # same names, with labels 2, 3, 4
```

### Q4 (medium). On `orders`: print the rows where the amount OR the items value is missing. Then print the number of missing values in every column.

```python
amt = orders["amount"].isna()
itm = orders["items"].isna()
print(orders[amt | itm])

print(len(orders.isna()))
print(orders.isna().sum())
```

### Q5 (medium). On `shop`: print the items whose name ends with `g`, then the items whose name has exactly 3 letters, then the items containing `OT` ignoring capitals.

```python
g = shop["item"].str.endswith("g")
print(shop.loc[g,"item"])
three = shop["item"].str.len()==3
print(shop.loc[three,"item"])
ot = shop["item"].str.contains("OT", case=False)
print(shop.loc[ot, "item"])               # Notebook, Bottle
```

### Q6 (hard). On `orders`: (a) print order_id and customer for Indore orders with amount above 1000, using `loc`. (b) Print `orders[~(orders['amount'] > 1000)]` and count its rows; explain in a comment why order 1002 is there. (c) Fix it so only orders with a real amount of 1000 or less remain, and count those rows.

```python
a = orders.query("city == 'Indore' & amount>1000")
a = orders.loc[(orders["city"] == "Indore") & (orders["amount"] > 1000), ["order_id", "customer"]]
print(a)    # 1001 Asha, 1004 Meena
# or in query style, then pick the columns:
print(orders.query('city == "Indore" and amount > 1000')[["order_id", "customer"]])

b = orders[~(orders["amount"] > 1000)]
print(b)           # 5 rows, order 1002 is among them
print(len(b))      # 5
# why 1002 is there: its amount is empty (NaN); NaN > 1000 is False, and ~False is True, so it sneaks in.



```

### Q7 (hard). On `shop`: keep items with a rating of 4 or more (empty ratings are not included) from the categories Stationery or Kitchen. Do it with a mask, then with `query`, and check that both give the same table with `equals`. Then renumber the labels from 0 and print the average price of the result. Last, in a comment, say what error `students.query('Marks > 80')` gives and why.

```python
# keep items with rating >= 4 (empty ratings excluded) from Stationery or Kitchen
m = shop[shop["rating"].notna() & (shop["rating"] >= 4) & shop["category"].isin(["Stationery", "Kitchen"])]   # mask version
q = shop.query('rating >= 4 and category in ["Stationery", "Kitchen"]')                                        # query version
print(m)                                   # Pen and Bottle
print(m.equals(q))                         # True: both ways give the same table

fresh = m.reset_index(drop=True)           # renumber labels from 0 (Pen was 0, Bottle was 2)
print(fresh)                               # labels 0 and 1
print(fresh["price"].mean())               # 130.0  ((10 + 250) / 2)

# students.query('Marks > 80') -> UndefinedVariableError: name 'Marks' is not defined
# the column is "marks"; capitals must match exactly in query too.
```
