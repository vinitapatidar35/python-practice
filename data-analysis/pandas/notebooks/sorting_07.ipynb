```python
import pandas as pd
```

### Key points: sort_values (Block 1)

- `df.sort_values("marks")` sorts the rows by one column, low to high by default.
- `ascending=False` gives high to low, such as a topper list.
- It returns a new table; `df` keeps its old order. Save the result.
- Each row keeps its own label, so after sorting the labels are out of order (2, 3, 0, 1).
- For fresh labels use `.reset_index(drop=True)` or `sort_values(..., ignore_index=True)`.
- After sorting, position and label differ: `iloc[0]` is the first row in the new order, `loc[0]` is the row that has label 0.
- Text sorts alphabetically, A to Z. Capital letters come before small letters.
- Sorting a single column (a Series) gives just its values, with their labels kept.
- `df["marks"].sort_values(ascending=False).head(2)` gives the top two marks.
- A wrong column name gives KeyError.

```python
s = pd.read_csv("students.csv")
```

```python
df = s.copy()

print(df.sort_values("marks"))
print(df.sort_values("marks", ascending=False))

top = df.sort_values("marks", ascending=False)
print(top.reset_index(drop=True))
print(df.sort_values("marks", ascending=False, ignore_index=True))

print(df.sort_values("name")["name"].tolist())
print(df["marks"].sort_values().tolist())
print(df["marks"].sort_values(ascending=False).head(2))

r = df.sort_values("marks")
print(r.iloc[0]["name"], r.loc[0, "name"])                  # position 0 is Ravi, label 0 is Asha: the labels moved with the rows
print(list(df["name"]) == list(r["name"]))       
```

### Key points: sorting by several columns (Block 2)

- `df.sort_values(["marks", "name"])` sorts by several columns. They go inside one list.
- The first column in the list is the main rule. The next one only settles ties in the first.
- The order of the list matters: `["city", "marks"]` and `["marks", "city"]` give different tables.
- `ascending=[False, True]` gives one direction per column, in the same order as the list. The list lengths must match, or ValueError.
- A single `ascending=False` applies to every column in the list.
- Ties that no column settles can come out in any order, so add a column until the order is fully decided.
- `ignore_index=True` or `reset_index(drop=True)` renumbers the labels after sorting.
- A top-N list is sort, then `head(N)`: `df.sort_values(["marks", "name"], ascending=[False, True]).head(3)`.
- A wrong column name inside the list gives KeyError.

```python
exam = pd.DataFrame({"name": ["Asha", "Vivek", "Ravi", "Meena", "Kiran", "Dev"],     # six students, three pairs with the same marks
                     "city": ["Indore", "Pune", "Delhi", "Indore", "Pune", "Delhi"],
                     "marks": [82, 91, 67, 82, 91, 67]})

print(exam.sort_values(["marks", "name"], ascending=[False, True]))
print(exam.sort_values(["marks","name"], ascending=False))

print(exam.sort_values(["name","marks"], ascending=False)["name"].tolist())

print(exam.sort_values(["marks", "name"], ascending=[False, True], ignore_index=True))
```

```python
shop = pd.DataFrame({"item": ["pen", "doll", "novel", "rice", "ball"],
                     "price": [10, 250, 400, 60, 120],
                     "rating": [4.5, None, 3.8, None, 4.2]})

print(shop.sort_values("rating"))
print(shop.sort_values("rating", ascending=False))
print(shop.sort_values("rating", ascending=False, ignore_index=True ))
print(shop.sort_values("rating", na_position="first"))
print(shop.dropna(subset=["rating"]).sort_values("rating", ascending=False))

print(shop.nlargest(2,"price"))
print(shop.nsmallest(2, "price"))                                    # the 2 rows with the smallest price, smallest first
print(shop.nlargest(2, "rating"))  
print(shop["price"].nlargest(2))

print(shop.nlargest(2, "price").equals(shop.sort_values("price", ascending=False).head(2)))
print(shop["rating"].sort_values().tolist())  

```

### Key points: sort_index and rank (Block 4)

- `df.sort_index()` sorts the rows by their labels, not by the values inside. `ascending=False` reverses it.
- After `sort_values` the labels are out of order. `sort_index()` puts the original order back, as long as the labels were the original ones.
- With a text index (`set_index("name")`), `sort_index()` gives an A to Z list by label.
- `sort_index(axis=1)` sorts the column names instead of the rows.
- `df["marks"].rank(ascending=False)` gives each row a position number. It moves no rows.
- Rank 1 is the smallest value by default. Use `ascending=False` to make the highest value rank 1.
- `rank` returns decimals, so add `.astype(int)` when there are no ties and no missing values.
- Tie handling is set by `method=`: `average` (default), `min`, `dense` or `first`.
- A rank column is usually followed by `sort_values("rank")` to get the rank list.
- `sort_values` moves rows; `rank` only labels them.

```python
s = pd.read_csv("students.csv") 
df = s.copy()

a = df.set_index("name")
print(a)
print(a.sort_index())
print(a.sort_index(ascending=False))

b = df.sort_values("marks")  
print(b)
print(b.index.tolist())
print(b.sort_index().equals(df))

print(df["marks"].rank(ascending=False).tolist())


r = df.copy()
r["rank"] = r["marks"].rank(ascending=False).astype(int)
print(r)
print(r.sort_values("rank"))

exam = pd.DataFrame({"name": ["Asha", "Vivek", "Ravi", "Meena", "Kiran", "Dev"],        # ties: Vivek and Kiran both have 91
                     "city": ["Indore", "Pune", "Delhi", "Indore", "Pune", "Delhi"],
                     "marks": [82, 91, 67, 82, 91, 67]})

e = exam.copy()                                                      # copy, so exam itself is not changed

e["avg"] = e["marks"].rank(ascending=False)                          # new column "avg": highest mark = rank 1
e["min"] = e["marks"].rank(ascending=False, method="min")            # new column "min"
e["dense"] = e["marks"].rank(ascending=False, method="dense")        # new column "dense"
e["first"] = e["marks"].rank(ascending=False, method="first")        # new column "first"

print(e)                                                             # show the table with all four rank columns
```
