### Q5 (medium). On a copy of `orders`: write a function `size(a)` that returns `big` if the amount is above 1000, `medium` if above 500, else `small`. Apply it to `amount` into a column `size`. Then fix the empty amount: use `loc` and `isna()` to set `size` to `unknown` for it. Print `order_id`, `amount`, `size` and the `value_counts()` of `size`.

```python
o = orders.copy()

def size(a):
    if a > 1000 :
        return "big"
    elif a > 500:
        return "medium" 
    else:
        return "small"

o["size"] = o["amount"].apply(size)
o.loc[o["amount"].isna(), "size"] = "unknown" 
print(o[["order_id","amount","size"]])
print(o["size"].value_counts())
```

### Q6 (hard). Build this table in code: `names = ["  priya SHARMA ", "ROHAN  gupta", "Anil-Mehta", "kavya rao"]` and `emails = ["Priya@Gmail.com", " ROHAN@yahoo.com", "anil@gmail.com ", "Kavya@Outlook.com"]`. Clean the names to `Priya Sharma` style (strip, title, dash to space, double space to one), clean the emails (strip, lower), add `first`, `last`, and `domain` columns, then print the `value_counts()` of `domain`. Last, print the raw `names` list to show the original is unchanged.

```python
names = ["  priya SHARMA ", "ROHAN  gupta", "Anil-Mehta", "kavya rao"]
emails = ["Priya@Gmail.com", " ROHAN@yahoo.com", "anil@gmail.com ", "Kavya@Outlook.com"]
raw = pd.DataFrame({"name": names, "email": emails})
df = raw.copy()

df["name"] = df["name"].str.strip()                           # spaces at the ends
df["name"] = df["name"].str.title()                           # Priya Sharma
df["name"] = df["name"].str.replace("-", " ", regex=False)    # dash to space
df["name"] = df["name"].str.replace("  ", " ", regex=False)   # TWO spaces to ONE

df["email"] = df["email"].str.strip().str.lower()             # clean the email first

parts = df["name"].str.split(" ", expand=True)                # pieces in columns 0 and 1
df["first"] = parts[0]
df["last"] = parts[1]
df["domain"] = df["email"].str.split("@").str[1]              # the part after @

print(df)
print(df["domain"].value_counts())
print(names)                                                  # the original list is unchanged

```

### Q7 (hard). On a copy of `orders`: (a) add `per_item`; set it to `0` where it is NaN, using `loc` and `isna()`. (b) Write a function `tier(row)` that returns `bulk` when `items` is 3 or more AND `amount` is above 1000, `repeat` when `items` is 1 or 2, otherwise `other`; apply it with `axis=1` into `tier`. (c) Print `customer`, `items`, `amount`, `tier`. (d) Using `query`, print the customers with `tier` equal to `bulk`. (e) In a comment, say why Meena's order (1004) gets the tier it does.

```python
o = orders.copy()

# (a) per_item, then repair the NaN rows with a mask
o["per_item"] = o["amount"] / o["items"]                   # NaN for orders 1002 and 1004
o.loc[o["per_item"].isna(), "per_item"] = 0                # loc + isna() writes 0 into those rows

# (b) a rule that needs several columns, so it gets a whole row
def tier(row):
    if row["items"] >= 3 and row["amount"] > 1000:
        return "bulk"
    elif row["items"] <= 2:
        return "repeat"
    else:
        return "other"

o["tier"] = o.apply(tier, axis=1)                          # axis=1 -> the rule runs on each ROW

# (c) the four columns
print(o[["customer", "items", "amount", "tier"]])

# (d) query for the bulk customers
print(o.query('tier == "bulk"')["customer"].tolist())      # ['Asha', 'Arjun']

# (e) Meena's items is NaN. NaN >= 3 is False, so she is not bulk, and NaN <= 2 is False,
#     so she is not repeat either. She falls into the else branch: other.
#     Her amount (2300) is big, but the rule never gets a chance to use it.
```
