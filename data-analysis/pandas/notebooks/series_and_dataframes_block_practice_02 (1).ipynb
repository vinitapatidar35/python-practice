## Block 1 — Building a DataFrame from a dict

### Block 1 — Theory: building a DataFrame from a dict

**Analogy:** a cupboard with labelled drawers. Each drawer has a label (`name`, `city`, `marks`) and holds a list of things. `pd.DataFrame` lines the drawers up side by side, and each drawer becomes one column.

**Why it matters:** you can make a small table in seconds without a file. It is how most practice tables and test data are built.

**Two Python words first**

| Word | Looks like | Meaning |
|---|---|---|
| list | `[82, 91, 67]` | values in order, inside square brackets |
| dict | `{"name": "Asha", "marks": 82}` | label to value pairs, inside curly brackets; the label is called a key |

```python
student = {"name": "Asha", "marks": 82}     # a dict: key -> value pairs
student["marks"]                             # look up a value by its key: 82
```

**Building a table**

```python
data = {
    "name":  ["Asha", "Vivek", "Ravi"],      # key "name" -> list of names
    "city":  ["Indore", "Pune", "Delhi"],    # key "city" -> list of cities
    "marks": [82, 91, 67],                   # key "marks" -> list of marks
}
df = pd.DataFrame(data)                      # turn the dict into a table
```

```
    name    city  marks
0   Asha  Indore     82
1  Vivek    Pune     91
2   Ravi   Delhi     67
```

- Each key becomes a column name, and each list fills that column.
- The first item of every list forms row 0, the second forms row 1, and so on.
- The row labels 0, 1, 2 are added automatically.

**Remember**

- All lists must have the same length, because every row needs a value in every column.
- Python is case-sensitive: it is `pd.DataFrame` with a capital D and F.

**Watch out**

- Lists of different lengths give `ValueError: All arrays must be of the same length`.
- `pd.dataframe(...)` (lowercase) gives `AttributeError: module 'pandas' has no attribute 'dataframe'`.
- Give tables descriptive names (`books`, not `x`).

## Block 2 — What a Series is

### Block 2 — Theory: what a Series is

**Analogy:** if a DataFrame is the whole cupboard, a Series is one drawer pulled out. It is a single column that keeps its row labels, so you still know which value belongs to which row.

**Why it matters:** most work happens on one column at a time (an average, a count). Every column you pick from a table is a Series, and in machine learning the column you want to predict (the target) is a Series.

| What you see | Name |
|---|---|
| the whole table | DataFrame |
| one column of it | Series |

```python
marks = pd.Series([82, 91, 67], name="marks")   # a Series: values + row labels + a name
marks
```

```
0    82
1    91
2    67
Name: marks, dtype: int64
```

```python
print(type(students))             # <class 'pandas.core.frame.DataFrame'>
print(type(students["marks"]))    # <class 'pandas.core.series.Series'>
```

**What a Series holds**

- values, row labels (the index), an optional name, and one dtype
- a DataFrame is several Series side by side that share the same row labels

**Why a Series, not a plain list?** The labels stay attached. After sorting or filtering, you still know which row each value came from. A plain list loses this. Column tools such as `.mean()`, `.describe()` and `.value_counts()` also live on a Series.

**Remember**

- A Series has only a length: `marks.shape` gives `(3,)`, one number, because it is one column.
- `type(x)` shows whether `x` is a Series, a DataFrame or something else.

**Watch out**

- Write `name=`: `pd.Series([82, 91], "marks")` puts the word in the row-labels slot and gives a TypeError.
- It is `pd.Series` with a capital S (`pd.series` gives AttributeError).
- You rarely type a Series by hand. Most come out of a DataFrame when you pick a column.

## Block 3 — Selecting one column

### Block 3 — Theory: selecting one column

**Analogy:** calling for a drawer by its label: "give me the marks drawer."

**Why it matters:** almost everything starts with "take this column": find its average, clean it, use it as the ML target.

```python
students["marks"]     # brackets: always works

```

```

1    91
2    67
```

**When the dot shortcut fails**

| Column name | Dot way | Brackets |
|---|---|---|
| has a space: `first name` | `table.first name` is broken code | `table["first name"]` works |
| same as a pandas action: `count` | gives the action, not the column | `table["count"]` works |

- Column names must match exactly, including capital letters and spaces.

- `students["Marks"]` gives `KeyError: 'Marks'` when the real name is `marks`. A KeyError means no column has that exact name.
- A hidden space in a header (`Price `) also causes a KeyError. Check `df.columns`.
- `survey.head(5)` shows the whole table, not one column: select first, then peek.

## Block 4 — Selecting many columns

### Block 4 — Theory: selecting many columns

**Analogy:** handing over a shopping list: "give me these items." To get several columns, you hand pandas a list of column names.

**Why it matters:** you rarely need all 85 survey columns. In ML this is how you choose your input columns (the features).

```python

```

```
    name  marks
0   Asha     82
1  Vivek     91
2   Ravi     67

```

```python
cols = ["marks", "name"]        # the list can be kept in a variable first
```

| Code | Gives |
|---|---|
| `students["marks"]` (one pair of brackets) | a Series |

- Forgetting the inner brackets, `students["name", "marks"]`, gives `KeyError: ('name', 'marks')`: pandas looks for one column with that tuple as its name.

## Block 6 — `iloc` (by position)

### Block 6 — Theory: `iloc` (selecting by position)

**Analogy:** floors in an Indian building. The ground floor is floor 0, the next is 1. A building with 4 floors has floors 0, 1, 2, 3, so there is no floor 4.

**Why it matters:** `iloc` lets you grab rows (and cells) by their position, no matter what the row labels are.

| You write | You get |
|---|---|
| `iloc[0]` | one row (a Series, sideways) |
| `iloc[-1]` | the last row (minus counts from the end) |
| `iloc[[0, 2]]` | chosen rows (a DataFrame) |
| `iloc[0:2]` | a range of rows; the stop is not included |
| `iloc[2, 1]` | one cell: row first, then column |

```python
students.iloc[0]          # the first row
students.iloc[[0, 2]]     # rows at positions 0 and 2
students.iloc[0:2]        # positions 0 and 1 (the stop, 2, is not included)
students.iloc[2, 1]       # row position 2, column position 1: 'Delhi'
```

**Remember**

- Positions count from 0, so the last of 4 rows is position 3 (or `-1`).
- `iloc` stands for integer location: position numbers only.
- `iloc[0:3]` gives the same rows as `head(3)`.
- `iloc[row, column]` always puts the row first.

**Watch out**

- `iloc` uses square brackets. `students.iloc(0)` does not give a row.
- A position past the end gives `IndexError: single positional indexer is out-of-bounds`.

## Block 7 — `loc` (by label)

### Block 7 — Theory: `loc` (selecting by label)

**Analogy:** calling a student by the name tag on the desk, not by where they sit. `iloc` asks "who sits in seat 2?"; `loc` asks "who has the name tag 2?"

**Why it matters:** after filtering, sorting or `tail()`, labels and positions no longer match. You need to know which one you are asking for.

```python
students.loc[0, "name"]                     # row label 0, column NAME: 'Asha'
students.loc[[0, 2], ["name", "marks"]]     # chosen row labels and chosen column names
students.loc[0:2]                           # labels 0 to 2: the stop IS included (3 rows)
```

**`iloc` versus `loc`**

| | `iloc` | `loc` |
|---|---|---|
| Finds rows by | position (counting from 0) | label (the tag on the left) |
| Columns by | number: `iloc[0, 2]` | name: `loc[0, "marks"]` |
| Slice `0:2` | stop excluded: 2 rows | stop included: 3 rows |
| Missing | IndexError | KeyError |

**The trap:** after `last_two = students.tail(2)`, the labels stay 2 and 3 but the positions are 0 and 1.

```python
last_two.loc[2]     # by label: Ravi
last_two.iloc[0]    # by position: Ravi
last_two.loc[0]     # KeyError: no row is labelled 0
```

**Remember**

- With the default index, labels equal positions; after `tail()`, filtering or sorting, they differ.
- `loc` takes column names, never column numbers.

**Watch out**

- `students.loc[0, 1]` gives `KeyError: 1`: use `students.loc[0, "city"]`.

### Practice

## Block 8 — `value_counts()`

### Block 8 — Theory: `value_counts()`

**Analogy:** counting votes after an election. You make a tally: Indore got 2 votes, Pune 1, Delhi 1.

**Why it matters:** it answers "what values does this column hold, and how often?" In the survey: people per country, hobbyists versus not. In ML it checks whether the target classes are balanced.

```python
students["city"].value_counts()      # select the column first, then count each value
```
```

Name: count, dtype: int64
```

```python
students["city"].value_counts(normalize=True)            # shares (they add up to 1)
survey["Country"].value_counts().head(5)                 # chain: count, then keep the top 5
survey["ConvertedComp"].value_counts(dropna=False)       # also count the missing (NaN) cells
```

**Remember**

- The result is a Series: the values are the labels, the counts are the numbers, biggest first.
- `normalize=True` gives shares; multiply by 100 for percentages.
- By default, NaN is ignored; `dropna=False` counts it too, so you can see how many people skipped a question.
- It works best on columns with few different values (categories).

**Watch out**

- `value_counts` works on one column: `students.value_counts("Pune")` gives `KeyError: 'Pune'`, because Pune is a value inside the column, not a column name.
- On a column like salary, nearly every value is different, so the tally is not useful.
- This is not the same as `count` in `describe()`: that is the total of filled cells; `value_counts` splits the total by value.
