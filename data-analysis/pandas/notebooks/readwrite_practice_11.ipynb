```python
import pandas as pd

shop = pd.DataFrame({"item":  ["pen", "bag", "lamp", "desk", "book"],
                     "city":  ["Indore", "Pune", None, "Delhi", "Pune"],
                     "price": [20, 450, 300, 2500, None],
                     "qty":   [10, 3, 5, 1, 7]})

# a messy file, written here so you do not need to download anything
with open("raw_sales.csv", "w") as f:
    f.write("report made by Vivek\n")
    f.write("code,date,amount\n")
    f.write("007,2026-10-01,500\n")
    f.write("012,2026-10-03,-\n")
    f.write("105,2026-10-09,300\n")
    f.write("230,2026-10-12,NA_\n")
print(shop)
```

Save `shop` into a file `shop.csv` (no extra index column). Read it back into `s1`. Print `s1.shape`, the list of column names, and whether `s1` is equal to `shop`. Then save it again WITHOUT `index=False` into `shop_bad.csv`, read it back into `bad`, and print its column names.

### Q2 (medium). Read a messy file

The file `raw_sales.csv` (made in the setup cell) has a junk first line, ids like 007 that must stay as written, a date column, and the words `-` and `NA_` for missing amounts. Read it into `r2` so that: the junk line is skipped, `code` stays text (007 must not become 7), `date` is a real date, and `-` and `NA_` become NaN. Print `r2`, `r2.dtypes`, and how many amounts are missing.

```python
# your code here (add a comment saying why for each line)
r2 = pd.read_csv("raw_sales.csv",
                 skiprows=1,                    # skip the junk first line
                 dtype={"code": str},           # keep 007 as text
                 parse_dates=["date"],          # text -> real dates
                 na_values=["-", "NA_"])        # these words mean missing
print(r2)
print(r2.dtypes)
print(r2["amount"].isna().sum())
```

### Q3 (medium to hard). Save a smaller table, read part of it

From `shop`, keep only the rows with `price` above 100, sorted by `price` from high to low, and only the columns `item` and `price`. Save this into `top.csv` with `;` as the separator (no index column). Read it back correctly into `t3` and print it. Then read only the first 2 rows of the file into `t3a` and print them. Finally print the column names you get if you forget the separator when reading.

```python
# your code here (add a comment saying why for each line)
top = shop[shop["price"] > 100].sort_values("price", ascending=False)[["item", "price"]]   # filter, sort, pick columns
top.to_csv("top.csv", index=False, sep=";")        # save with ; separator
t3 = pd.read_csv("top.csv", sep=";")               # read with the SAME separator
print(t3)
t3a = pd.read_csv("top.csv", sep=";", nrows=2)     # only the first 2 rows
print(t3a)
print(pd.read_csv("top.csv").columns.tolist())     # forgot sep
```
