### Text to dates
- CSV dates arrive as text (`str`/`object`). Check `df.info()`.
- `df["d"] = pd.to_datetime(df["d"])` makes real dates (`datetime64`). Save it back.
- Real dates can be compared, sorted, `min()`/`max()`, and filtered: `df[df["d"] > "2026-10-01"]`.
- Standard text is year-month-day (`2026-10-05`). For other styles give `format`: `format="%d/%m/%Y"` (%d day, %m month, %Y 4-digit year, %y 2-digit, %b Oct, %B October, %H:%M time).
- Without `format`, `10/05/2026` is read month first (10 Oct). Day-first data needs `format="%d/%m/%Y"`.
- `errors="coerce"` turns bad dates into `NaT` (the date NaN). `isna()`, `dropna()`, `fillna()` work on it.
- Without `coerce`, one bad entry gives ValueError.
- A date from columns: `pd.to_datetime(df[["year","month","day"]])` (column names must be exactly year, month, day).
- A date with no time shows `00:00:00`.
- Watch out: forgetting the conversion (`.dt` then gives AttributeError); wrong `format` gives ValueError or wrong dates.

```python
import pandas as pd

t = pd.DataFrame({"trip":   [1, 2, 3, 4],
                  "start":  ["2026-10-01", "2026-10-05", "2026-09-28", "2026-10-12"],
                  "amount": [500, 300, 450, 800]})
t.info()                                   # start is text (str / object)

t["start"] = pd.to_datetime(t["start"])    # text -> real dates
t.info()                                   # start is datetime64
print(t)

print(t["start"].min())                    # earliest date
print(t["start"].max())                    # latest date
print(t.sort_values("start"))              # sorted by date
print(t[t["start"] > "2026-10-01"])        # filter by date (compare with plain text)

# other date styles
print(pd.to_datetime("15/10/2026", format="%d/%m/%Y"))          # 2026-10-15
print(pd.to_datetime("10/05/2026"))                             # read month first: 2026-10-05
print(pd.to_datetime("10/05/2026", format="%d/%m/%Y"))          # day first: 2026-05-10
s = pd.Series(["15/10/2026", "01/09/2026", "31/12/2025"])
print(pd.to_datetime(s, format="%d/%m/%Y"))
print(pd.to_datetime("15 Oct 2026"), pd.to_datetime("October 15, 2026"))

# bad dates
bad = pd.Series(["2026-10-01", "hello", "2026-10-03"])
print(pd.to_datetime(bad, errors="coerce"))     # hello -> NaT

# a date from separate columns
d = pd.DataFrame({"day": [15, 16], "month": [10, 10], "year": [2026, 2026]})
print(pd.to_datetime(d))

# fail on purpose (remove the # to try one at a time)
# pd.to_datetime(bad)                           # ValueError: "hello" does not match the format
# pd.Series(["2026-10-01"]).dt.year             # AttributeError: .dt needs real dates
```

### Parts of a date (.dt)
- Needs real dates (`pd.to_datetime` first), else AttributeError.
- No brackets: `.dt.year`, `.month`, `.day`, `.hour`, `.minute`, `.dayofweek` (Monday = 0 ... Sunday = 6), `.quarter`, `.dayofyear`.
- With brackets: `.dt.day_name()`, `.dt.month_name()`.
- Save pieces as columns (features for ML): `df["month"] = df["d"].dt.month`.
- Weekend: `df["d"].dt.dayofweek >= 5`.
- Filter by a piece: `df[df["d"].dt.month == 10]`.
- Filter by a range: `df[(df["d"] >= "2026-10-01") & (df["d"] < "2026-11-01")]`.
- `between(a, b)` includes both ends; a text date means midnight at the start of that day.
- Group by a piece: `df.groupby(df["d"].dt.month)["x"].sum()`. Month names sort A to Z, so use the number for order.
- Back to text: `df["d"].dt.strftime("%d-%m-%Y")`; `.dt.normalize()` sets the time to 00:00.
- Watch out: `.dt.month()` with brackets is a TypeError; `.dt` on a text column is an AttributeError.

```python
import pandas as pd

t = pd.DataFrame({"trip":   [1, 2, 3, 4, 5, 6],
                  "start":  ["2026-10-01 09:30", "2026-10-05 18:45", "2026-09-28 07:10",
                             "2026-10-12 14:00", "2026-09-13 22:15", "2026-11-02 11:20"],
                  "amount": [500, 300, 450, 800, 250, 600]})
t["start"] = pd.to_datetime(t["start"])
print(t)

# step 1: pieces of a date
print(t["start"].dt.year.tolist())
print(t["start"].dt.month.tolist())
print(t["start"].dt.day.tolist())
print(t["start"].dt.hour.tolist())
print(t["start"].dt.minute.tolist())
print(t["start"].dt.day_name().tolist())
print(t["start"].dt.month_name().tolist())
print(t["start"].dt.dayofweek.tolist())     # Monday = 0 ... Sunday = 6
print(t["start"].dt.quarter.tolist())

# step 2: save pieces as columns
t["year"] = t["start"].dt.year
t["month"] = t["start"].dt.month
t["weekday"] = t["start"].dt.day_name()
t["is_weekend"] = t["start"].dt.dayofweek >= 5
print(t)

# step 3: filter
print(t[t["start"].dt.month == 10])                                       # October trips
print(t[t["is_weekend"]])                                                 # weekend trips
print(t[t["start"].dt.hour >= 12])                                        # afternoon and evening
print(t[(t["start"] >= "2026-10-01") & (t["start"] < "2026-11-01")])      # whole of October
print(t[t["start"].between("2026-09-20", "2026-10-05")])                  # inclusive range

# step 4: group by a piece
print(t.groupby("month")["amount"].sum())
print(t.groupby(t["start"].dt.month_name())["amount"].sum())
print(t.groupby("weekday")["amount"].count())
print(t["start"].dt.month.value_counts().sort_index())

# step 5: back to text
print(t["start"].dt.strftime("%d-%m-%Y").tolist())
print(t["start"].dt.strftime("%b %Y").tolist())
print(t["start"].dt.normalize().tolist())

# fail on purpose (remove the # to try one at a time)
# pd.Series(["2026-10-01"]).dt.year      # AttributeError: still text
# t["start"].dt.month()                  # TypeError: month has no brackets
```

```python
o = pd.DataFrame({"order_id":  [1, 2, 3, 4, 5],
                  "ordered":   ["2026-10-01", "2026-10-03", "2026-09-25", "2026-10-10", "2026-09-30"],
                  "delivered": ["2026-10-04", "2026-10-10", "2026-09-27", None, "2026-10-02"],
                  "amount":    [500, 300, 450, 800, 250]})
o["ordered"] = pd.to_datetime(o["ordered"])
o["delivered"] = pd.to_datetime(o["delivered"])      # None becomes NaT (a missing date)
print(o)
```

```python
gap = o["delivered"] - o["ordered"]
print(gap)
print(gap.dt.days)
```

```python
o["days_taken"] = (o["delivered"] - o["ordered"]).dt.days
print(o)
```

```python
o["days_taken"].mean()            # 3.5  -> NaN is skipped
o["days_taken"].max()             # 7.0
o[o["days_taken"] > 3]            # slow deliveries: order 2
```

```python
o["due"] = o["ordered"] + pd.Timedelta(days=7)
print(o["due"])
```
