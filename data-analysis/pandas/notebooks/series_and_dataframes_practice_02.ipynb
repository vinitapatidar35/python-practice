### Q16

What are the dtypes of `amount` and `items` in `orders`? Why are they not `int64`?

```python

print(orders["amount"].dtype)    # float64
print(orders["items"].dtype)     # float64
# amount: order 1002 has an empty cell; items: order 1004 has an empty cell.
# An empty cell is stored as NaN, and NaN is a decimal, so the whole column becomes float64, not int64.
```

### Q17

What happens with `shop["qty"] + 1`? Explain why.

```python
print(shop["qty"].dtype)    # object -> the column is stored as text
shop["qty"] + 1             # TypeError: can only concatenate str (not "int") to str
# qty has the text cell "five", so the WHOLE column is object. Text + number is not allowed.
```

### Q20

Without running it first, explain the difference between `orders[["city"]]`, `orders["city"]` and `orders.loc[:, "city"]`: the type and the shape of each. Then run all three to check.

```python
a = orders[["city"]]          # list inside brackets -> DataFrame (a table with one column)
b = orders["city"]            # one name -> Series (one column)
c = orders.loc[:, "city"]     # all rows (:), column named "city" -> Series, same as b

print(type(a), a.shape)       # <class 'pandas.core.frame.DataFrame'> (8, 1)
print(type(b), b.shape)       # <class 'pandas.core.series.Series'> (8,)
print(type(c), c.shape)       # <class 'pandas.core.series.Series'> (8,)
```
