# Topic 6A practice: get_dummies

The tables are built in code, so no CSV is needed. Re-run the setup cell to reset them. Start each question from `students` or `shop`, and give each result a new name.

Tip: a comment line must start with `#`.

````python
import pandas as pd

students = pd.DataFrame({"name": ["Asha", "Vivek", "Ravi", "Meena"],
                         "city": ["Indore", "Pune", "Delhi", "Indore"],
                         "marks": [82, 91, 67, 75]})

shop = pd.DataFrame({"item": ["pen", "doll", "novel", "rice", "ball"],
                     "category": ["Stationery", "Toys", "Books", "Food", "Toys"],
                     "price": [10, 250, 400, 60, 120]})
print(students)
print(shop)
````

### Q1 (easy). The first encoding

Turn the `city` column of `students` into 0/1 columns (use `dtype=int`). Save in `d1`. Print `d1` and the list of its column names.

````python
# your code here (add a comment saying why for each line)

s = students.copy()
d1 = pd.get_dummies(s["city"],dtype=int)
print(d1)
print(list(d1.columns))
````

### Q2 (easy). Add a prefix

Do the same, but make the new column names start with `city_`. Save in `d2` and print the column names.

````python
# your code here (add a comment saying why for each line)
s = students.copy()
d1 = pd.get_dummies(s["city"],dtype=int, prefix="city")
print(d1)
print(list(d1.columns))
````

### Q3 (easy to medium). Read the boxes

Using `d1`-style boxes (without the prefix), print the column totals, then print the sum of each row as a list. Say in a comment what each of the two results means.

````python
# your code here (add a comment saying why for each line)
d = pd.get_dummies(students["city"], dtype=int)   # one 0/1 column for every city

print(d.sum())                                    # column totals: how many students live in each city (Delhi 1, Indore 2, Pune 1)
print(d.sum(axis=1).tolist())                     # row sums: each student ticks exactly one box, so every row is 1
````

### Q4 (medium). Encode inside the whole table

Encode only the `city` column of `students` and keep `name` and `marks`. Save in `s4`. Print `s4`, `s4.shape` and `students.shape`.

````python
# your code here (add a comment saying why for each line)
s = students.copy()
s4 = pd.get_dummies(s,columns=["city"])
print(s4.shape)
print(students.shape)
````

### Q5 (medium). drop_first

Encode the `category` column of `shop` with `drop_first=True` (keep the other columns, `dtype=int`). Save in `s5`. Print the column list. Which category has no box? Then print the item that belongs to it, using only the boxes (a mask where all category boxes are 0).

### Q7 (hard). Train and test

Use `students` as `train` and this `test`:

```python
test = pd.DataFrame({"name": ["Kiran", "Dev"], "city": ["Goa", "Indore"], "marks": [88, 70]})
```

1. Encode `city` in both (`dtype=int`) and print whether their column lists are equal.
2. Fix the test table so it has exactly the train columns, with missing boxes as 0.
3. Print the fixed test table and say in a comment what happened to Kiran's Goa.

````python
# your code here (add a comment saying why for each line)
test = pd.DataFrame({"name": ["Kiran", "Dev"], "city": ["Goa", "Indore"], "marks": [88, 70]})   # the new batch; Goa never appears in train

tr = pd.get_dummies(students, columns=["city"], dtype=int)    # train boxes: city_Delhi, city_Indore, city_Pune
te = pd.get_dummies(test, columns=["city"], dtype=int)        # test boxes: city_Goa, city_Indore (only what it sees)
print(list(tr.columns) == list(te.columns))                   # False: the two tables do not line up

fixed = te.reindex(columns=tr.columns, fill_value=0)          # force exactly the train columns; any missing box is filled with 0
print(fixed)                                                  # Kiran has 0 in every city box: Goa was never seen in train, so it gets no box
print(list(fixed.columns) == list(tr.columns))                # True: now the layouts match
````
