### Block 2 — Reading a CSV file
- CSV (Comma-Separated Values) is a table stored as plain text.
- The first line is the header: the column names.
- Every other line is one row; commas separate the columns.
- `pd.read_csv(path)` reads a CSV file and returns a DataFrame.
- Store the result in a variable (usually `df`) to use it later.
- A variable alone on the last line of a cell is shown as a neat table (notebooks only).
- `print(df)` shows the table as plain text and works on any line and in .py files.
- After editing a CSV file, run `read_csv` again; the DataFrame does not update by itself.

### Block 5 — shape and info()
- `df.shape` gives the size of the table as (rows, columns).
- `df.info()` prints one line per column: name, non-null count and data type, plus the total rows and memory used.
- Non-null count = how many values are filled in; fewer than the number of rows means the column has missing values.
- Common dtypes: int64 = whole numbers, float64 = decimals, object = text.
- Attributes (like `shape`) have no brackets; methods (like `info()`) have brackets because they perform an action.
- Jupyter shows only the value of the last line in a cell; use `print()` to show values from earlier lines.
- `info()` prints by itself, so its report shows from any line.
- Run `shape` and `info()` first on every new dataset.

```python
df = pd.read_csv("students.csv")
df.shape
print(df.shape)
df.info()                           # one line per column: name, filled count, data type
#Rule to remember: to see several things from one cell, use print() for everything except the last line.
```

### Block 6 — head() and tail()
- `df.head(n)` shows the first n rows; `df.tail(n)` shows the last n rows.
- With no number, both show 5 rows.
- Row labels keep their original numbers; they are not renumbered.
- Rows are counted from 0, so the last row's label is (number of rows - 1).
- They return a small copy for viewing; the original DataFrame is unchanged.
- Use them to see real values in a large table without printing everything.

### Block 7 — Display options
- `pd.set_option(name, value)` changes how much pandas shows; the data itself never changes.
- Settings start with `pd.`, so they apply to every table in the notebook, not one table.
- `display.max_columns`: most columns shown before the middle is hidden with "..." (default 20). None = no limit.
- `display.max_rows`: most rows shown before the middle is hidden (default 60). Use a number, never None.
- `display.max_colwidth`: most characters shown per text cell (default 50). None = full text.
- The value is a limit: a table bigger than the limit is still cut with "...".
- `pd.reset_option(name)` puts one setting back to its default; restarting the kernel resets all settings.
- Put settings in a cell right after `import pandas as pd`.
- For very large tables, use head() and tail() instead of raising max_rows.

### Block 8 — describe()
- `df.describe()` gives summary statistics for every number column.
- Text columns are skipped by default.
- count = how many values are filled in (missing values are not counted).
- mean = average; std = how spread out the values are.
- min and max = smallest and largest values.
- 25%, 50%, 75% = values a quarter, half and three quarters of the way up the sorted data; 50% is the middle value (median).
- Use it to see typical values and to spot strange ones, such as impossible minimums or maximums.
