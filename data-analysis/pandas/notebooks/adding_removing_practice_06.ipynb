The tables below are built in code, so no CSV is needed. Re-run the setup cell to reset them. Work on copies (`.copy()`) so the originals stay safe.

### Q7 (hard). Mix everything

On a copy of `orders`:

1. Remove the `amount` column and keep it as a Series `amt` (one line).
2. Put it back as the 2nd column (position 1) with the name `amount`.
3. Drop the `customer` column.
4. Move `city` to the front without typing the other names.
5. Add a new row `Pune, 1005, 1100` using the column order you now have.

Print the final table and its column list.

````python
o = orders.copy()                                          # copy first, because pop and insert change the table itself
amt = o.pop("amount")                                      # step 1: remove amount from o and keep its values in amt
print(list(o.columns))                                     # ['order_id', 'customer', 'city']

o.insert(1, "amount", amt)                                 # step 2: put it back at position 1 (insert is in place, no "o =")
print(list(o.columns))                                     # ['order_id', 'amount', 'customer', 'city']

o = o.drop(columns=["customer"])                           # step 3: drop returns a new table, so save it back into o
print(list(o.columns))                                     # ['order_id', 'amount', 'city']

others = [c for c in o.columns if c != "city"]             # step 4a: every column except city, in the original order
o = o[["city"] + others]                                   # step 4b: city first, then the others
print(list(o.columns))                                     # ['city', 'order_id', 'amount']

o.loc[len(o)] = ["Pune", 1005, 1100]                       # step 5: values in the CURRENT column order: city, order_id, amount
print(o)
print(list(o.columns))                                     # ['city', 'order_id', 'amount']
````
