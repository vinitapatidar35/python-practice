```python
import pandas as pd                                    # open the pandas toolbox
pd.set_option("display.max_columns", None)             # show every column in every table

students = pd.read_csv("students.csv")                 # name, city, marks (4 rows)
orders = pd.read_csv("orders.csv")                     # order_id, customer, city, amount, items (8 rows)
```

### What the index is

- The index is the set of row labels, shown on the left of a table
- It is not a data column, and `shape` does not count it
- `df.index` holds the row labels (attribute, no brackets); `df.columns` holds the column labels
- A CSV gets default labels 0, 1, 2 ... (a RangeIndex)
- Labels stay with their rows when the table is cut; positions restart from 0

```python
print(students.index)          # RangeIndex(start=0, stop=4, step=1) -> the row labels
print(list(students.index))    # [0, 1, 2, 3] -> as a normal list
print(students.index[0])       # 0 -> the label at position 0
print(students.shape)          # (4, 3) -> only 3 columns: the index is NOT counted as a column
print(students.index.name) 
```

### set_index

- `df.set_index("col")` makes a column the row labels instead of 0, 1, 2 ...
- It returns a new table; the original is unchanged, so save the result in a variable
- The column moves out of the data into the index: it is no longer in `columns`, and `shape` loses one column
- `df.index.name` holds the name of the column that became the index
- Pick a column with a different value in every row (an id or a name)
- A wrong column name gives KeyError

```python
o = orders.set_index("order_id")    # make the order_id column the row labels (returns a NEW table)
print(o)                            # order_id now sits on the left, not among the columns
```

```python
print(o.index)
print(o.shape)
print(list(o.columns))
print(o.index.name) 
print(orders)
print(orders.index)
print(orders.index.name)
```

### loc with a custom index

- After `set_index("col")`, `loc` finds rows by the values of that column: `o.loc[1003]`
- `loc[label, "column"]` gives one value; `loc[a:b]` is a range with the stop included
- A list of labels (`loc[[1001, 1005]]`) gives separate rows
- Text labels need quotes: `c.loc["Ravi", "city"]`
- `iloc` still works by position; label and position are different numbers
- A missing label gives KeyError; a position out of range with iloc gives IndexError

```python
o = orders.set_index("order_id")     # order_id is now the row label
```

```python
print(o.loc[1001])
print(o.loc[1001:1003])
print(o.loc[[1001,1003]])
print(o.loc[1002,"city"])
```

```python
o = orders.set_index("order_id")
r = o.reset_index()
print(r)
print(r.index.name)
print(r.index.shape)
print(r.head(3))
```

### reset_index

- `df.reset_index()` puts the row labels back as a normal column and renumbers the rows 0, 1, 2 ...
- `df.reset_index(drop=True)` renumbers the rows and throws the old labels away
- It returns a new table; save the result in a variable
- A named index (order_id) becomes a column with that name; an unnamed index becomes a column called "index"
- Use it to undo `set_index`, or to renumber a table that was cut by head, tail or filtering

```python
# ---- Demo 1: set_index, then reset_index (the old labels are kept as a column) ----
by_name = students.set_index("name")          # name becomes the row labels; it leaves the columns
print(by_name)

restored = by_name.reset_index()              # name comes back as a normal column; rows renumbered 0, 1, 2 ...
print(restored)

# ---- Demo 2: a cut table, then reset_index(drop=True) (the old labels are thrown away) ----
last_two = students.tail(2)                   # keeps the last 2 rows with their old labels 2 and 3
print(last_two)

renumbered = last_two.reset_index(drop=True)  # throws the old labels away; rows renumbered 0, 1
print(renumbered)

# ---- Demo 3: the danger of drop=True on real data ----
print(by_name.reset_index(drop=True))         # names are LOST here, because they were the labels
```

### sort_index

- `df.sort_index()` sorts the rows by their labels (the index), not by column values
- Text labels sort A to Z; number labels sort by size
- `df.sort_index(ascending=False)` reverses the order
- It returns a new table; save the result in a variable
- Sorting changes positions but each row keeps its label
- To sort by column values use `sort_values` (Topic 7)

```python
name = students.set_index("name")
a = name.sort_index()
print(a)
b = name.sort_index(ascending=False)
print(b)
```

```python
o = orders.set_index("order_id")             # order ids as labels
print(o.sort_index(ascending=False).head(3)) # biggest order id first, top 3 rows
```

```python
sorted = name.sort_index()
print(name.iloc[1].name)       # Vivek -> position 1 BEFORE sorting
print(sorted.iloc[1].name)     # Meena -> position 1 AFTER sorting
```

### Duplicate labels and index.name

- Index labels are allowed to repeat (unlike a database primary key)
- `df.loc[label]` returns a DataFrame if several rows share the label, a Series if only one does
- `df.index.is_unique` is True when every label appears once
- `df.index.name` holds the index's name (None for the default 0, 1, 2 ...); it can be changed by assignment
- `reset_index()` turns a named index into a column with that name
- Choose a unique column (an id) as the index for clean lookups

```python
c = orders.set_index("city")
print(c)
```

```python
print(c.loc["Indore"])        # 3 rows share the label -> a DataFrame (a table of 3 rows)
print(c.loc["Mumbai"])        # 1 row has the label    -> a Series (a single row)
print(c.loc["Indore", "amount"])   # a column of 3 values -> a Series of 3 numbers
```

```python
print(c.index.is_unique)      # False -> some labels repeat
o = orders.set_index("order_id")
print(o.index.is_unique)      # True  -> every label appears once
```
