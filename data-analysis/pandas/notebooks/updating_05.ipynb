## Block 1: Changing values with loc

- `df.loc[row, column] = value` writes a value into a table.
- The rows part can be a label (`2`) or a mask (`df["city"] == "Indore"`); the mask changes every matching row at once.
- The column part is a column name in quotes.
- Assigning to a column name that does not exist creates a new column; rows not matched by the mask get NaN.
- Writing changes the table in place. It does not return a new table, so there is no need to save the result in a variable.
- To keep the original safe, make a separate copy first: `df = students.copy()`.
- Without `.copy()`, `df = students` gives a second name for the same table, so changing `df` also changes `students`.
- Writing needs `loc` with row and column together. `df[mask]["marks"] = 0` often changes nothing, because `df[mask]` is a temporary copy.
- A single `=` writes, `==` compares.

```python
import pandas as pd                                    # open the pandas toolbox
students = pd.read_csv("students.csv")                 # name, city, marks (4 rows)
df = students.copy()                                   # a separate copy, so students stays untouched

df.loc[2, "marks"] = 70                                # row label 2, column marks -> Ravi's mark becomes 70
print(df)

df.loc[df["city"] == "Indore", "city"] = "Indore City" # every row where the mask is True gets the new city text
print(df)

df.loc[df["marks"] >= 80, "grade"] = "A"               # a column that doesn't exist yet is created; unmatched rows get NaN
print(df)

print(students)                                        # the original, unchanged
```

## Block 2: Column arithmetic and new columns

- A calculation on a column applies to every row at once: `df["marks"] + 5`, `df["price"] * 0.9`.
- Signs: `+`, `-`, `*`, `/` (normal division), `//` (division without the remainder), `%` (remainder), `**` (power).
- Two columns of the same table work row by row: `df["amount"] / df["items"]`.
- The result is a new Series; the table changes only when it is saved with `=`.
- `df["new"] = ...` makes a new column; if the name already exists, it overwrites that column.
- A single value (`df["country"] = "India"`) is copied into every row.
- A calculation involving NaN gives NaN in that row.
- `+` joins text columns, but text plus a number gives TypeError; a number cannot be joined to text this way.
- A whole column gives one number with `sum()`, `mean()`, `max()`, `min()`; the empty cells are skipped.
- Make a copy (`df = students.copy()`) before adding columns if the original table must stay unchanged.

## Block 3: rename and replace

- `df.rename(columns={"old": "new"})` changes column names; the dictionary is old name : new name.
- It gives a new table and leaves the original unchanged; save it with `=` to keep the change.
- A wrong or misspelled old name is ignored silently: no error, no change. Check with `list(df.columns)`.
- `df.rename(columns=str.upper)` applies a function to every column name.
- `df.rename(index={0: "a"})` renames row labels in the same way.
- `df.columns = ["a", "b", "c"]` replaces all names at once; the list must have exactly one name per column, or ValueError.
- `df["col"].replace("old", "new")` changes one value inside one column; `replace({"old1": "new1", "old2": "new2"})` changes several.
- On the whole table use `df.replace({"column": {"old": "new"}})`.
- `replace` matches the whole cell, not a part of the text.
- `replace` gives a new Series or table; assign it back to change the data: `df["col"] = df["col"].replace(...)`.
- A column holding a mix of text and numbers stays `object` after replacing one value; changing the type comes later with `astype`.




```python
df = students.copy()                                       # a copy to work on

r = df.rename(columns={"name": "student", "marks": "score"})   # old name : new name, inside a dictionary
print(list(r.columns))                                     # ['student', 'city', 'score']
print(list(df.columns))                                    # ['name', 'city', 'marks'] -> df is unchanged

print(list(df.rename(columns=str.upper).columns))          # a function applied to every name: ['NAME', 'CITY', 'MARKS']
print(df.rename(columns={"Name": "x"}).columns.tolist())   # wrong capital: no error, nothing changes

d2 = df.copy()
d2.columns = ["n", "c", "m"]                               # replace ALL names at once; needs exactly 3 names
print(list(d2.columns))                                    # ['n', 'c', 'm']

print(df["city"].replace("Indore", "Indore City"))         # one value -> another, inside one column
print(df["city"].replace({"Indore": "IDR", "Pune": "PNQ"}))   # many at once with a dictionary
print(df.replace({"city": {"Delhi": "New Delhi"}}))        # on the whole table: {column: {old: new}}

shop = pd.read_csv("shop.csv")
shop["qty"] = shop["qty"].replace("five", 5)               # fix the stray text cell from Topic 2
print(shop["qty"].tolist(), shop["qty"].dtype)             # ['50', '20', 5, '12', '3', '1'] object
```

## Block 5: apply

- `df["col"].apply(function)` runs a function on every value of a column and returns a new Series.
- The function takes one value and gives one value back. Write the function name with no brackets: `apply(grade)`.
- `def name(x): ... return ...` makes your own rule; `return` is what comes back for each value.
- `lambda x: x + 5` is a one-line function with no name; it is good for short rules.
- `df.apply(function, axis=1)` runs the function on each ROW; the function receives the whole row, and you read values with `row["col"]`.
- `axis=0` (the default for the table) gives each column to the function; for a rule on rows, always write `axis=1`.
- The result is a new Series; save it with `=`: `df["grade"] = df["marks"].apply(grade)`.
- NaN values are passed into the function too; `NaN > 1000` is False, so the else branch runs. Handle it if empty cells matter.
- `apply` is slower than column arithmetic. For `+`, `*`, `/` and comparisons use the column directly; use `apply` when the rule needs if/else or several steps.
- `map(dictionary)` is for lookups, `apply(function)` is for rules.
- `apply(grade())` with brackets fails (`grade() missing 1 required positional argument`); pass the function itself.

```python
df = students.copy()                                   # a copy to work on

def grade(m):                                          # your own rule: takes ONE mark, gives ONE grade back
    if m >= 80:
        return "A"                                     # 80 or more -> A
    elif m >= 70:
        return "B"                                     # 70 to 79 -> B
    else:
        return "C"                                     # everything else -> C

df["grade"] = df["marks"].apply(grade)                 # run the rule on every mark; no brackets after grade
print(df)

print(df["marks"].apply(lambda m: m + 5).tolist())     # a one-line rule with no name (lambda): [87, 96, 72, 80]

df["intro"] = df.apply(lambda row: row["name"] + " from " + row["city"], axis=1)   # axis=1 -> the rule gets a whole ROW
print(df["intro"].tolist())                            # ['Asha from Indore', 'Vivek from Pune', ...]

print(df["marks"].apply(grade).value_counts())         # apply gives a Series, so value_counts works: A 2, B 1, C 1

orders = pd.read_csv("orders.csv")
def size(a):
    if a > 1000:
        return "big"
    else:
        return "small"
print(orders["amount"].apply(size).tolist())           # the empty amount (order 1002) becomes 'small' - NaN > 1000 is False
```

## Block 6: String cleaning with .str

- Text tools need `.str` in front: `df["name"].str.strip()`. `df["name"].strip()` fails with AttributeError.
- `str.strip()` removes spaces (and new lines) from the start and end, not from the middle. `lstrip()` and `rstrip()` do one side only.
- `str.lower()`, `str.upper()` and `str.title()` change the capitals.
- `str.replace("old", "new", regex=False)` changes part of each text; `regex=False` treats the text as plain characters.
- `str.split(" ")` cuts each text at a separator; the result is a list per row.
- `str.split(" ", expand=True)` puts the pieces in separate columns, numbered 0, 1, 2 ...
- `.str[1]` takes piece number 1 from a list; `.str[:3]` takes the first 3 letters of a text. The same bracket works on both.
- Tools can be chained in order: `df["email"].str.strip().str.lower()`; each `.str` tool works on the output of the one before.
- Every one of these returns a new Series; save it with `=` to change the table.
- Clean first, then compare or group: `" Pune"` and `"Pune"` are different values until stripped.
- On a column holding NaN, the result is NaN for those rows; `.str` on a column of numbers gives AttributeError.

```python
raw = pd.DataFrame({
    "name":  ["  asha SHARMA ", "VIVEK  singh", "Ravi-Kumar", "meena patel"],       # extra spaces, mixed capitals, a dash
    "email": ["Asha@Gmail.com", "VIVEK@yahoo.com", " ravi@gmail.com ", "meena@Outlook.com"],
})
c = raw.copy()                                                    # work on a copy

c["name"] = c["name"].str.strip()                                 # remove spaces at the start and end
c["name"] = c["name"].str.title()                                 # first letter of each word in capitals
c["name"] = c["name"].str.replace("-", " ", regex=False)          # replace a dash with a space (as plain text)
c["name"] = c["name"].str.replace("  ", " ", regex=False)         # two spaces -> one space
print(c["name"].tolist())                                         # ['Asha Sharma', 'Vivek Singh', 'Ravi Kumar', 'Meena Patel']

parts = c["name"].str.split(" ", expand=True)                     # cut at each space; expand=True gives separate columns
c["first"] = parts[0]                                             # the first piece
c["last"] = parts[1]                                              # the second piece

c["email"] = c["email"].str.strip().str.lower()                   # chain two washes: strip then lower
c["domain"] = c["email"].str.split("@").str[1]                    # cut at @, then take piece number 1
print(c["first"].str[:3].tolist())                                # the first 3 letters of each: ['Ash', 'Viv', 'Rav', 'Mee']
print(c)
print(raw["name"].tolist())                                       # raw is unchanged (still messy)
```
