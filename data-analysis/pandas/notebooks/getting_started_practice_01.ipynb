**Q13 (Real data).** Run `survey.describe()`: (a) median `Age`? (b) maximum `Age`, is it realistic? (c) How many people did not report a salary (`ConvertedComp`)? Do the subtraction in code.

````python
survey.describe()                                   # summary of the 6 number columns
````

````python
print("median Age:", survey["Age"].median())      # the 50% row of describe()
print("max Age:", survey["Age"].max())              # the highest age reported
total = len(survey)                                 # total number of people (rows)
reported = survey["ConvertedComp"].count()          # people who reported a salary (non-missing)
print("no salary:", total - reported)               # people who skipped the question
````

**Answer:**

- (a) The median Age is the **50%** row of `Age` (about 29 in the original dataset; check your own output). Half of the people who gave an age are that age or younger.
- (b) The maximum Age is **99**. A 99-year-old developer in an online survey is very unlikely: a joke or a typo. A value far from the rest is an **outlier**, and `describe()` is how you spot it.
- (c) Missing = total rows minus the `ConvertedComp` count. It is roughly 33 thousand people (about 37%). These are NaN values: they skipped the question.

**Q14 (Debug story).** A friend adds a new row to `shop.csv` and saves. (1) `shop.shape` still shows the old size. (2) After restarting the kernel, `shop.shape` gives `NameError: name 'shop' is not defined`. Explain both and write the cell that fixes both.

````python
shop = pd.read_csv("shop.csv")      # fresh copy of the file: includes the new row, and shop exists again
shop.shape                           # now shows the new number of rows
````

**Answer:**

1. `shop` is a **snapshot** (copy) made when `read_csv` ran. Editing the file later does not change that copy, so `shape` shows the old size.
2. Restarting the kernel **wipes memory**, so the variable `shop` no longer exists. `NameError` means Python does not know that name.

Fix for both: run `read_csv` again.
