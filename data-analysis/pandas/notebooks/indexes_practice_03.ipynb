### Q6 (hard). On `shop`: set `item` as the index (`by_item`), sort it Z to A by label, and print the first item name using `.index[0]`. Then bring `item` back as a column in a new table `back` and print `list(back.columns)`. Last: what do you lose with `by_item.reset_index(drop=True)`?

### Q7 (hard). On `orders` with `city` as the index (`c`): (a) print `c.index.is_unique`; (b) print the type of `c.loc['Mumbai']` and of `c.loc['Indore']`; (c) print the total amount for Indore using `loc` and `sum()`; (d) set `order_id` as an index (`o`), print `o.index.is_unique`, and rename its index to `id`. Say in a comment which of city/order_id is a good key and why.

```python
c = orders.set_index("city")             # city table
print(c.index.is_unique)                 # False
print(type(c.loc["Mumbai"]))             # Series
print(type(c.loc["Indore"]))             # DataFrame
print(c.loc["Indore", "amount"].sum())   # total Indore amount

o = orders.set_index("order_id")         # separate table, separate name
o.index.name = "id"
print(o.index.name)                      # id
print(o.index.is_unique)                 # True
# comment: which is the good key, city or order_id, and why?
```
