2. Create a null vector (an array of all zeros) of size 10.

```python
null_vector = np.zeros(10)
print(null_vector)
```

3.Create the same null vector of size 10, but this time make the 5th value equal to 1.

```python
null_vector = np.zeros(10)
null_vector[4] = 1
print(null_vector)
```

4.Create a vector with values ranging from 10 to 49 (inclusive of 10, and you can decide whether 49 is included based on what function you use).

```python
arr = np.arange(10,50)
print(arr)
```

5.Take the vector you just created and reverse it (so it goes from 49 down to 10).

```python
arr = arr[::-1]
print(arr)
```

6.Create a 3x3 matrix with values ranging from 0 to 8.

```python
arr = np.arange(0,9).reshape(3,3)
print(arr)
```
