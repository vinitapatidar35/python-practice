## np.nan and np.inf

**np.nan (Not a Number)**
- Represents a missing or undefined value (e.g. the result of `0/0`)
- Very common in Pandas when data has gaps
- `nan` is contagious: any arithmetic with `nan` gives `nan`
- `nan` is never equal to anything, not even itself

**np.inf (Infinity)**
- Represents a value larger than any number
- Any comparison involving `nan` is always False

**Floating-point precision**
- Computers store decimals approximately, so exact `==` comparisons on floats can fail
- To compare floats safely, use `np.isclose(a, b)` instead of `==`

## Floating-Point Precision

**The problem**
- Decimal arithmetic on computers is slightly inexact:
  - `0.1 + 0.2` gives `0.30000000000000004`, not `0.3`
  - `0.1 + 0.2 == 0.3` gives `False`

**Why it happens**
- Computers store numbers in binary
- Some decimals (like 0.1) can't be represented exactly in binary, just like 1/3 can't be written exactly in decimal (0.333...)
- The computer stores a very close approximation, and tiny errors add up in calculations

**The fix**
- Never use `==` to compare floats
- Use `np.isclose(a, b)`, which returns `True` if the two numbers are within a tiny tolerance
- Example: `np.isclose(0.1 + 0.2, 0.3)` gives `True`

**Good to know**
- Integers are always exact (`2 + 3 == 5` is always `True`)
- The issue only affects decimal (float) numbers

````python
print(0 * np.nan) # nan
print(np.nan == np.nan) # False
print(np.inf > np.nan) # False, Every comparison involving nan (>, <, ==) returns False
print(np.nan - np.nan) #nan
print(0.3 == 3 * 0.1) # False
````

## np.diag()

**What it does**
- Builds a square matrix by placing the given values along a diagonal and filling everything else with zeros
- Syntax: `np.diag(values, k=0)`

**The main diagonal**
- The line of positions from the top-left corner to the bottom-right corner of a matrix

**The `k` parameter (which diagonal to use)**
- `k=0`: main diagonal (default)
- `k>0`: diagonals above the main one (`k=1` is one step above)
- `k<0`: diagonals below the main one (`k=-1` is one step below)

**Matrix size**
- `size = number of values + |k|`
- Example: 3 values with `k=0` gives a 3x3 matrix; 3 values with `k=1` or `k=-1` gives a 4x4 matrix
- Diagonals away from the main one are shorter, so the matrix grows to make room for all the values

**Examples**
```python
np.diag([1, 2, 3])          # 3x3, values on the main diagonal
np.diag([1, 2, 3], k=1)     # 4x4, values one step above the main diagonal
np.diag([1, 2, 3], k=-1)    # 4x4, values one step below the main diagonal
```

**Also good to know**
- If you pass a 2D matrix instead of a list, `np.diag()` does the reverse: it extracts the diagonal values from it

17.Create a 5x5 matrix of random values and normalize it, meaning subtract the mean and divide by the standard deviation.

````python
normalize = np.random.random((5,5))
mean = normalize.mean()
std = np.std(normalize)

final_output = (normalize - mean) / std
print(final_output)
print(final_output.mean())    # should be very close to 0
print(final_output.std())     # should be very close to 1
````

## np.argmax() and np.argmin()

**What they do**
- `np.argmax(arr)` returns the **position (index)** of the largest value in an array
- `np.argmin(arr)` returns the **position (index)** of the smallest value in an array
- They give a position, not the value itself

**max vs argmax (value vs position)**
- `arr.max()` gives the largest **value**
- `np.argmax(arr)` gives the **index** where that value sits

**Points to remember**
- Give `argmax` the whole array, not a single value
- If the maximum appears more than once, `argmax` returns the position of the **first** one
- For a 2D array, `argmax` works on the flattened array unless you pass `axis`: `axis=0` gives the position of the maximum in each column, `axis=1` in each row


````python
vector = np.random.random(10)    # 10 lockers, each with a random number
a = np.argmax(vector)            # a = the locker NUMBER (position) that holds the biggest value
vector[a] = 0                    # open that locker and put 0 inside
print(vector)                    # the biggest value is gone, replaced by 0
````
